# 4. Loading Our First Small Language Model

```python
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM
)

import torch
```

```python
MODEL_NAME = "Qwen/Qwen2.5-0.5B-Instruct"

print(MODEL_NAME)
```

```python
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

print("Tokenizer Loaded Successfully")
```

```python
print(type(tokenizer))
```

# 5. Loading the SLM into GPU Memory

```python
gpu = torch.cuda.get_device_properties(0)

print("="*60)

print("GPU Information")

print("="*60)

print(f"Name : {gpu.name}")

print(f"Total Memory : {gpu.total_memory/1024**3:.2f} GB")

print(f"Multiprocessors : {gpu.multi_processor_count}")

print("="*60)
```

```python
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.float16,
    device_map="auto"
)

print("Model Loaded Successfully")
```

# 6. Our First Inference

```python
prompt = """
What is Artificial Intelligence?
"""
```

```python
inputs = tokenizer(prompt, return_tensors="pt")
```

```python
print(inputs)
```

```python
print(tokenizer.decode(inputs["input_ids"][0]))
```

# 7. Running Our First Inference

```python
inputs = tokenizer(
    prompt,
    return_tensors="pt"
)

inputs = inputs.to(model.device)

print(f"Model Device : {model.device}")
print(f"Input Device : {inputs['input_ids'].device}")
```

```python
with torch.no_grad():

    outputs = model.generate(
        **inputs,
        max_new_tokens=150,
        temperature=0.7,
        do_sample=True
    )
```

```python
response = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print(response)
```

# 9. Understanding Embeddings

```python
embedding_layer = model.model.embed_tokens

print(embedding_layer)
```

```python
embedding_weights = embedding_layer.weight

print(embedding_weights.shape)
```

```python
print(f"Shape : {embedding_weights.shape}")

print(f"Total Parameters : {embedding_weights.numel():,}")

print(f"Memory : {embedding_weights.numel()*2/1024/1024:.2f} MB")
```

```python
word = "dog"
```

```python
tokens = tokenizer(word)

print(tokens)
```

```python
token_id = tokens["input_ids"][0]

vector = embedding_weights[token_id]

print(vector.shape)
```

```python
print(vector[:20])
```
